# Функция для быстрого просмотра столбца

```python
def my_analis(data_set, column, figsize=(10, 5)):
    nan_percent = data_set[column].isna().mean() * 100
    nan_count = data_set[column].isna().sum()
    count_unique = data_set[column].nunique()
    top_values = data_set[column].value_counts(dropna=False, ascending=False).head(count_unique if count_unique <= 10 else 10)
    types = data_set[column].apply(type).value_counts()
    print(f"1) Процент пустых записей: {nan_percent}")
    print(f"2) Количество пустых записей: {nan_count}")
    print(f"3) Количество уникальных записей: {count_unique}")
    print(f"4) Топ уникальных записей для {top_values}:")
    print(f"5) Количество типов записей: {types}")
    with sns.axes_style("whitegrid"):
        plt.figure(figsize=figsize)
        sns.barplot(
            y=top_values.values, 
            x=top_values.index)


my_analis(df, "loyalty_program_type_nm", (15, 5))
```

#### Анализ среднего чека и количества заказов по ДНЯМ для отелей (HOT) и авиабилетам (AIR) без учёта 1% самых дорогих заказов

```python
hot_cnt = df2_HOT_interval.groupby('date').size().reset_index(name='cnt')
hot_avg = df2_HOT_interval.groupby('date')['nominal_price_rub_amt'].mean().reset_index(name='avg')

air_cnt = df2_AIR_interval.groupby('date').size().reset_index(name='cnt')
air_avg = df2_AIR_interval.groupby('date')['nominal_price_rub_amt'].mean().reset_index(name='avg')

hot_daily = hot_cnt.merge(hot_avg, on='date', how='outer').sort_values('date')
hot_daily['cnt_pct'] = hot_daily['cnt'] / hot_daily['cnt'].max() * 100
hot_daily['avg_pct'] = hot_daily['avg'] / hot_daily['avg'].max() * 100

air_daily = air_cnt.merge(air_avg, on='date', how='outer').sort_values('date')
air_daily['cnt_pct'] = air_daily['cnt'] / air_daily['cnt'].max() * 100
air_daily['avg_pct'] = air_daily['avg'] / air_daily['avg'].max() * 100


plt.figure(figsize=(12,5))
sns.lineplot(data=hot_daily, x='date', y='cnt_pct', label='Кол-во заказов, % от max')
sns.lineplot(data=hot_daily, x='date', y='avg_pct', label='Средний чек, % от max')
plt.title('Отели: количество успешных (SUC) заказов и средний чек заказа (нормировано)')
plt.xlabel('Дата')
plt.ylabel('% от максимума')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


plt.figure(figsize=(12,5))
sns.lineplot(data=air_daily, x='date', y='cnt_pct', label='Кол-во заказов, % от max')
sns.lineplot(data=air_daily, x='date', y='avg_pct', label='Средний чек, % от max')
plt.title('Авиабилеты: количество успешных (SUC) заказов и средний чек заказа (нормировано)')
plt.xlabel('Дата')
plt.ylabel('% от максимума')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Для отелей явно прослеживается зависимость между пиками бронирований и средним чеком: при снижении стоимости бронирования резко взлетает количество бронирований. Для авиабилетов такая тенденция тоже наблюдается, однако она не так значительна**

#### Анализ среднего чека по МЕСЯЦАМ для отелей (HOT) и авиабилетам (AIR) без учёта 1% самых дорогих заказов

```python
hot_avg_month = df2_HOT_interval.groupby('month')['nominal_price_rub_amt'].mean().reset_index(name='avg_month')
hot_cnt_month = df2_HOT_interval.groupby('month').size().reset_index(name='cnt_month')

air_avg_month = df2_AIR_interval.groupby('month')['nominal_price_rub_amt'].mean().reset_index(name='avg_month')
air_cnt_month = df2_AIR_interval.groupby('month').size().reset_index(name='cnt_month')

hot_monthly = hot_avg_month.merge(hot_cnt_month, on='month', how='outer').sort_values('month')
hot_monthly['avg_per_month'] = hot_monthly['avg_month'] / hot_monthly['avg_month'].max() * 100
hot_monthly['cnt_per_month'] = hot_monthly['cnt_month'] / hot_monthly['cnt_month'].max() * 100

air_monthly = air_avg_month.merge(air_cnt_month, on='month', how='outer').sort_values('month')
air_monthly['avg_per_month'] = air_monthly['avg_month'] / air_monthly['avg_month'].max() * 100
air_monthly['cnt_per_month'] = air_monthly['cnt_month'] / air_monthly['cnt_month'].max() * 100

plt.figure(figsize=(12,5))
sns.lineplot(data=hot_monthly, x='month', y='cnt_per_month', label='Кол-во заказов, % от max')
sns.lineplot(data=hot_monthly, x='month', y='avg_per_month', label='Средний чек, % от max')
plt.title('Отели: количество успешных (SUC) заказов и средний чек заказа (нормировано)')
plt.xlabel('Месяц')
plt.ylabel('% от максимума')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,5))
sns.lineplot(data=air_monthly, x='month', y='cnt_per_month', label='Кол-во заказов, % от max')
sns.lineplot(data=air_monthly, x='month', y='avg_per_month', label='Средний чек, % от max')
plt.title('Авиабилеты: количество успешных (SUC) заказов и средний чек заказа  (нормировано)')
plt.xlabel('Месяц')
plt.ylabel('% от максимума')
plt.tight_layout()
plt.show()
```

**Можно предположить, что есть зависимость между средним чеком и количеством заказов за месяц для отелей. А вот для авиабилетов такой зависимости явно не наблюдается (по крайней мере по ммесяцам). Вероятно, это вызвано тем, что отели часто сами по себе являются целью путешествия (или же то место, где они находятся) из-за чего скидка на бронирование номера логично повышает количество бронирований. Тогда как авиабилет - лишь средство достижения (перелёта) для основного приобретения - отдыха в новом месте и заселения в отель. Поэтому изменение цены авиабилетов не сильно влияет на их покупаемость => привлекать клиентов скидками больший смысл имеет для отелей, чем для авиабилетов**
